# Import data clean

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/processed/telco_clean.csv', index_col='customerID')
df.head()
```

# Splitting dataset

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop(columns=['Churn'])
y = df["Churn"].map({"No": 0, "Yes": 1})

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
print('label train :', y_train.value_counts())
print('\n label test :', y_test.value_counts())
```

```python
print('label train:',(y_train.value_counts(normalize=True) * 100).round(2).astype(str) + '%')
print('label test:',(y_test.value_counts(normalize=True) * 100).round(2).astype(str) + '%')
```

# Simpan

```python
import joblib
```

```python
data_split = {
    "X_train": X_train,
    "X_test": X_test,
    "y_train": y_train,
    "y_test": y_test
}

joblib.dump(data_split, "../data/processed/data_split.pkl")
```
