# Cek Distribusi Data

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import math

# Memilih kolom kategorik
kolom_kategorik = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# Tidak menampilkan customerID karena hanya ID pelanggan
if "customerID" in kolom_kategorik:
    kolom_kategorik.remove("customerID")

# Mengatur jumlah grafik
n_kolom = 3
n_baris = math.ceil(len(kolom_kategorik) / n_kolom)

fig, axes = plt.subplots(
    n_baris, n_kolom,
    figsize=(18, n_baris * 4)
)

axes = axes.flatten()

# Membuat bar chart
for i, kolom in enumerate(kolom_kategorik):
    sns.countplot(
        data=df,
        x=kolom,
        order=df[kolom].value_counts().index,
        ax=axes[i]
    )

    axes[i].set_title(f"Distribusi {kolom}")
    axes[i].set_xlabel(kolom)
    axes[i].set_ylabel("Jumlah Pelanggan")
    axes[i].tick_params(axis="x", rotation=45)

    # Menambahkan angka pada setiap batang
    for container in axes[i].containers:
        axes[i].bar_label(container, fmt="%.0f", fontsize=8)

# Menghapus subplot kosong
for j in range(len(kolom_kategorik), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

Terlihat bahwa pada data kategorik datanya konsisten, tidak ada yg berbeda penulisan, typo , dll
