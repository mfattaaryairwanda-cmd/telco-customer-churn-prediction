# 01 — Business Understanding
## Customer Churn Prediction

---

## 1. Latar Belakang

Dalam industri telekomunikasi, mempertahankan pelanggan merupakan salah satu aspek penting dalam menjaga keberlangsungan bisnis. Pelanggan yang berhenti menggunakan layanan perusahaan atau *customer churn* dapat memengaruhi pendapatan dan meningkatkan kebutuhan perusahaan untuk memperoleh pelanggan baru.

Salah satu pendekatan yang dapat digunakan untuk memahami permasalahan tersebut adalah analisis data pelanggan dan pemodelan *machine learning*. Data pelanggan dapat memberikan gambaran mengenai karakteristik demografi, penggunaan layanan, jenis kontrak, serta biaya berlangganan yang berkaitan dengan status pelanggan.

Oleh karena itu, proyek ini dilakukan untuk membangun model klasifikasi yang dapat memprediksi status *customer churn* berdasarkan karakteristik pelanggan yang tersedia dalam dataset IBM Telco Customer Churn.

## 2. Identifikasi Masalah

Permasalahan yang menjadi fokus dalam proyek ini adalah:

1. Perusahaan perlu memahami karakteristik pelanggan yang berhenti berlangganan.
2. Data pelanggan mencakup berbagai aspek, seperti demografi, layanan telepon dan internet, kontrak, metode pembayaran, serta biaya berlangganan.
3. Diperlukan pendekatan berbasis data untuk mengidentifikasi pola yang berkaitan dengan status *churn*.
4. Berbagai algoritma klasifikasi perlu dievaluasi untuk mengetahui model yang memberikan performa prediksi paling sesuai.

## 3. Rumusan Masalah

Berdasarkan latar belakang tersebut, rumusan masalah dalam proyek ini adalah:

1. Bagaimana karakteristik pelanggan berdasarkan status *churn*?
2. Bagaimana hubungan antara karakteristik pelanggan, layanan yang digunakan, dan biaya berlangganan dengan status *churn*?
3. Bagaimana kinerja berbagai algoritma *machine learning* dalam memprediksi status *churn*?
4. Algoritma mana yang memberikan performa prediksi terbaik berdasarkan metrik evaluasi yang digunakan?

## 4. Tujuan Proyek

Tujuan dari proyek ini adalah:

1. Memahami karakteristik pelanggan berdasarkan data yang tersedia.
2. Melakukan eksplorasi dan pembersihan data sebelum pemodelan.
3. Menyiapkan data untuk proses pelatihan dan pengujian model.
4. Membangun model klasifikasi menggunakan beberapa algoritma *machine learning*.
5. Membandingkan performa model berdasarkan metrik evaluasi yang relevan.
6. Menginterpretasikan hasil analisis sebagai dasar penyusunan rekomendasi untuk mendukung strategi retensi pelanggan.

## 5. Dataset yang Digunakan

Proyek ini menggunakan dataset **IBM Telco Customer Churn** yang tersedia melalui Kaggle.

**Sumber dataset:**  
https://www.kaggle.com/datasets/blastchar/telco-customer-churn

Dataset terdiri dari 7.043 baris dan 21 kolom. Setiap baris merepresentasikan satu pelanggan, sedangkan setiap kolom berisi identitas atau karakteristik pelanggan.

Dataset mencakup informasi demografi, layanan yang digunakan, informasi akun, biaya berlangganan, dan status *churn*.

## 6. Target Prediksi

Target prediksi dalam proyek ini adalah variabel `Churn`, yang menunjukkan apakah pelanggan berhenti berlangganan atau tidak.

Target memiliki dua kategori:

- `Yes`: pelanggan melakukan *churn*.
- `No`: pelanggan tidak melakukan *churn*.

Untuk keperluan pemodelan, target akan dikonversi menjadi nilai numerik:

- `Yes` menjadi `1`.
- `No` menjadi `0`.

Dengan demikian, permasalahan ini termasuk dalam **klasifikasi biner (*binary classification*)**, yaitu memprediksi salah satu dari dua kelas berdasarkan karakteristik pelanggan.

## 7. Pendekatan Analisis

Proyek akan dilakukan melalui beberapa tahapan berikut.

1. **Business Understanding** — menentukan permasalahan, tujuan, dan manfaat analisis.
2. **Exploratory Data Analysis (EDA)** — memeriksa struktur data, distribusi variabel, komposisi kelas *churn*, dan pola awal pelanggan.
3. **Data Cleaning** — memeriksa tipe data, nilai kosong, duplikasi, serta konsistensi data.
4. **Data Preprocessing** — memisahkan fitur dan target, membagi data menjadi training dan testing, serta menyiapkan transformasi data yang diperlukan.
5. **Model Development** — melatih dan melakukan penyesuaian parameter beberapa algoritma klasifikasi.
6. **Model Evaluation** — mengevaluasi dan membandingkan performa model menggunakan data testing.
7. **Interpretation and Recommendation** — menginterpretasikan hasil analisis dan menyusun rekomendasi berdasarkan temuan yang diperoleh.

Algoritma yang direncanakan untuk dieksplorasi meliputi:

- Logistic Regression
- K-Nearest Neighbors (KNN)
- Decision Tree
- Random Forest
- Support Vector Machine (SVM)
- XGBoost

Setiap algoritma akan menggunakan pembagian data training dan testing yang konsisten agar perbandingan performa dapat dilakukan secara lebih adil. Penanganan nilai kosong, encoding variabel kategorikal, dan scaling variabel numerik akan disesuaikan dengan kebutuhan algoritma melalui pipeline preprocessing.

## 8. Metrik Evaluasi

Evaluasi model akan menggunakan beberapa metrik berikut.

| Metrik | Kegunaan |
|---|---|
| Accuracy | Mengukur proporsi seluruh prediksi yang benar |
| Precision | Mengukur ketepatan prediksi pelanggan yang diperkirakan melakukan *churn* |
| Recall | Mengukur proporsi pelanggan yang benar-benar melakukan *churn* dan berhasil teridentifikasi |
| F1-Score | Menggabungkan precision dan recall |
| ROC-AUC | Mengukur kemampuan model membedakan kelas *churn* dan non-*churn* berdasarkan skor prediksi |

Penggunaan beberapa metrik diharapkan dapat memberikan gambaran performa model yang lebih menyeluruh. Hal ini penting karena kesalahan dalam mengidentifikasi pelanggan yang berpotensi berhenti berlangganan dapat memiliki konsekuensi bisnis yang berbeda dari kesalahan prediksi pada pelanggan yang tidak melakukan *churn*.

## 9. Manfaat Proyek

Proyek ini diharapkan memberikan manfaat berikut.

1. **Manfaat analitis:** memberikan gambaran mengenai karakteristik pelanggan dan pola yang berkaitan dengan *churn*.
2. **Manfaat teknis:** menjadi sarana untuk mempelajari tahapan pengembangan model klasifikasi, mulai dari eksplorasi data hingga evaluasi model.
3. **Manfaat bisnis:** memberikan informasi awal yang dapat dipertimbangkan perusahaan dalam menentukan kelompok pelanggan yang memerlukan perhatian untuk strategi retensi.

Hasil model merupakan alat bantu pengambilan keputusan dan bukan jaminan bahwa pelanggan tertentu akan berhenti berlangganan.

## 10. Batasan Proyek

Batasan dalam proyek ini adalah sebagai berikut.

1. Analisis hanya menggunakan variabel yang tersedia dalam dataset IBM Telco Customer Churn.
2. Prediksi difokuskan pada status *churn*, bukan pada waktu pasti pelanggan berhenti berlangganan.
3. Hubungan antara fitur dan status *churn* tidak secara langsung menunjukkan hubungan sebab-akibat.
4. Performa model bergantung pada kualitas data, proses preprocessing, pembagian data, pemilihan algoritma, dan konfigurasi parameter.
5. Rekomendasi bisnis disusun berdasarkan hasil analisis dan tidak secara langsung dianggap sebagai strategi yang telah terbukti meningkatkan retensi pelanggan.

## 11. Kesimpulan Business Understanding

Proyek *Customer Churn Prediction* berfokus pada pemanfaatan data pelanggan telekomunikasi untuk memahami karakteristik dan pola yang berkaitan dengan keputusan pelanggan berhenti berlangganan.

Dengan memanfaatkan variabel demografi, informasi layanan, lama berlangganan, kontrak, metode pembayaran, serta biaya layanan, proyek ini akan mengembangkan dan mengevaluasi beberapa model klasifikasi untuk memprediksi status *churn*.

Tahap berikutnya adalah melakukan *Exploratory Data Analysis* (EDA) untuk memeriksa struktur, kualitas, distribusi, dan karakteristik awal dataset sebelum melanjutkan ke proses pembersihan data.
